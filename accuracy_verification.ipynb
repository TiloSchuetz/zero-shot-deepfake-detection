## Test accuracy of different sparsification approaches

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import spearmanr, kendalltau

import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

from sklearn.metrics import (
    f1_score,
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    mean_absolute_error
)
```

```python
fast = pd.read_csv("/home/tischuet/zero-shot-deepfake-detection/imagenet_results/Imagenet_val_metaclip2_fast.csv")
fast_sorted_score = fast.sort_values(by=["img_path"]).score

standard = pd.read_csv("/home/tischuet/zero-shot-deepfake-detection//imagenet_results/Imagenet_val_metaclip2_standard.csv")
standard_sorted_score = standard.sort_values(by=["img_path"]).score

fast_sparse = pd.read_csv("/home/tischuet/zero-shot-deepfake-detection//imagenet_results/Imagenet_val_metaclip2_fast_sparse.csv")
fast_sparse_sorted_score = fast_sparse.sort_values(by=["img_path"]).score

fast_batch_four = pd.read_csv("/home/tischuet/zero-shot-deepfake-detection//imagenet_results/Imagenet_val_metaclip_fast_batch4.csv")
fast_batch_four_sorted_score = fast_batch_four.sort_values(by=["img_path"]).score
```

```python
def verify(exact, approx):
    mea = mean_absolute_error(exact, approx)
    print(f"Mean absolute error: {mea}")

    rho, _  = spearmanr(exact, approx)
    print(f"Spearman rho: {rho}")

    tau, _  = kendalltau(exact, approx)
    print(f"Kendall tau: {tau}")

    return None
```

```python
verify(standard_sorted_score, fast_sorted_score)
```

```python
verify(fast_sorted_score, fast_batch_four_sorted_score)
```

```python
verify(standard_sorted_score, fast_sparse_sorted_score)
```

```python
verify(fast_sorted_score, fast_sparse_sorted_score)
```
